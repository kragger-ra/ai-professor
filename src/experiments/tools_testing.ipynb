```python
class SimpleCodeAgent:
    """Simple implementation of a code-executing agent using smolagents"""

    def __init__(self, ctx_swarm: dict, tools: Optional[List] = None, rag_model=None):
        if rag_model is None:
            rag_model = RagModel()
            print("[SMO RAG MODEL INIT]")
        self.mono_agent = MonologueAgent()

        self.ctx_swarm = ctx_swarm
        self.handler = CtxHandler(ctx_swarm)
        if tools is None:
            self.tools = setup_tools(
                self.ctx_swarm, mono_agent=self.mono_agent, ctx_handler=self.handler
            )
        # Initialize LLM model
        self.model = get_smo_llm()
        self.rag_model = rag_model
        self.running = False
        self.last_event_time = time.time()
```

```python
ctx_swarm = create_ctx_swarm()
agent = SimpleCodeAgent()
```
