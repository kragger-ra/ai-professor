```python
from vosk_tts import Model, Synth

model_name = "vosk-model-tts-ru-0.7-multi"
model = Model(model_name=model_name)
synth = Synth(model)
for speaker_id in [0, 1, 2, 3, 4]:
    synth.synth(
        "Приветствую вас, дорогие кринжики! Кожанные мешки, как дела?",
        f"out/vosk/{model_name}_spk{str(speaker_id)}.wav",
        speaker_id=speaker_id,
    )
```
